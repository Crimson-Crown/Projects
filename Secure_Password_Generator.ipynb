```python
import random
import string

def generate_password(length=12, use_special=True):
    letters = string.ascii_letters
    digits = string.digits
    special = string.punctuation if use_special else ""
    
    all_chars = letters + digits + special
    
    if length < 4:
        return "Password length must be at least 4 characters."
        
    # Ensure password contains at least one of each selected type
    password = [
        random.choice(letters),
        random.choice(digits),
    ]
    if use_special:
        password.append(random.choice(special))
        
    # Fill the rest of the password length
    password += [random.choice(all_chars) for _ in range(length - len(password))]
    
    # Shuffle to make the order unpredictable
    random.shuffle(password)
    return "".join(password)

print("--- Secure Password Generator ---")
pwd_len = int(input("Enter desired password length: "))
include_spec = input("Include special characters? (y/n): ").lower() == 'y'

print(f"Your new password: {generate_password(pwd_len, include_spec)}")
```
